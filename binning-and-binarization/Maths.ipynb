```python
from sklearn.preprocessing import KBinsDiscretizer, Binarizer
import numpy as np
x = np.array([10, 20, 30, 40, 50, 90]).reshape(-1, 1)
print(KBinsDiscretizer(n_bins=3, encode='ordinal', strategy='uniform').fit_transform(x).ravel())
print(KBinsDiscretizer(n_bins=3, encode='ordinal', strategy='quantile').fit_transform(x).ravel())
print(Binarizer(threshold=35).fit_transform(x).ravel())
```
