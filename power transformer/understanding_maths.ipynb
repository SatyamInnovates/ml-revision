```python
from sklearn.preprocessing import PowerTransformer
from scipy.stats import skew
import numpy as np

x = np.array([30000, 40000, 50000, 60000, 10000000]).reshape(-1, 1)
pt = PowerTransformer(method='yeo-johnson')
out = pt.fit_transform(x)

print(pt.lambdas_)        # is it below 1?
print(skew(x.ravel()), skew(out.ravel()))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import skew
from sklearn.preprocessing import PowerTransformer

x = np.array([30000, 40000, 50000, 60000, 10000000]).reshape(-1, 1)
out = PowerTransformer(method='yeo-johnson').fit_transform(x)

fig, ax = plt.subplots(1, 2, figsize=(10, 4))

ax[0].hist(x.ravel(), bins=10)
ax[0].set_title(f"Before (skew = {skew(x.ravel()):.2f})")

ax[1].hist(out.ravel(), bins=10)
ax[1].set_title(f"After (skew = {skew(out.ravel()):.2f})")

plt.show()
```
