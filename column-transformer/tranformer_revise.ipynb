```python
import pandas as pd
```

```python
df = pd.read_csv('covid_toy.csv')
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(df.drop(columns=['has_covid'],axis=1),
                                                df['has_covid'],
                                                test_size=0.2)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OrdinalEncoder,OneHotEncoder
```

```python
transformer = ColumnTransformer(transformers=[
    ('trf1',SimpleImputer(),['fever']),
    ('trf2',OrdinalEncoder(categories=[['Mild','Strong']]),['cough']),
    ('trf3',OneHotEncoder(sparse_output=False,drop='first'),['gender','city'])
],remainder='passthrough')
```

```python
x_train_transformer = transformer.fit_transform(x_train)
```

```python

x_train_transformer
```

```python
column_names = transformer.get_feature_names_out()
x_train_df = pd.DataFrame(x_train_transformer,columns=column_names)
```
