```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('titanic_toy.csv')
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.shape
```

```python
df['new_Age'] = df['Age'].fillna(df['Age'].median())
```

```python
fig = plt.figure()
ax = fig.add_subplot(111)

df['Age'].plot(kind='kde',ax=ax,color='red')
df['new_Age'].plot(kind='kde',ax=ax,color='green')
plt.show()
```

```python
print(df['Age'].isnull().mean() * 100)
```

```python
imputer = SimpleImputer(strategy='constant',fill_value=1)
df['new_Age'] = imputer.fit_transform(df[['Age']])
```

```python
fig = plt.figure()
ax = fig.add_subplot(111)

df['Age'].plot(kind='kde',ax=ax,color='red')
df['new_Age'].plot(kind='kde',ax=ax,color='green')
plt.show()
```

```python
imputer2 = SimpleImputer(strategy="most_frequent")
df['new_Age'] = imputer2.fit_transform(df[['Age']])
```

```python
fig = plt.figure()
ax = fig.add_subplot(111)

df['Age'].plot(kind='kde',ax=ax,color='red')
df['new_Age'].plot(kind='kde',ax=ax,color='green')
plt.show()
```
